```python
fig = px.pie(
    df,
    names="depression_label",
    hole=0.45,
    title="Depression Distribution"
)
fig.show()
```

```python


import seaborn as sns

sns.scatterplot(data=df,
                x='daily_social_media_hours',
                y='stress_level')

plt.title("daily_social_media_hours vs stress_level")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.hist(df['mental_health_risk_score'], bins=20, edgecolor='black')
plt.title("Mental Health Risk Score Distribution")
plt.xlabel("Risk Score")
plt.ylabel("Frequency")
plt.show()


```

```python
plt.figure(figsize=(10,8))

sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap='coolwarm')

plt.title("Correlation Heatmap")
plt.show()
```

```python
d = df.groupby('daily_social_media_hours')[['stress_level', 'anxiety_level']].mean().reset_index()

fig = px.line(
    data_frame=d,
    x='daily_social_media_hours',
    y='stress_level',
    markers=True,
    template='plotly_dark',
    color_discrete_sequence=['pink']
)

fig.show()
```

```python
fig = px.violin(df,
                x='daily_social_media_hours',
                y='age',
                box=True)
fig.show()
```

```python

    
fig = px.imshow(
    df.corr(numeric_only=True),
    text_auto=True,
    color_continuous_scale="Reds",
    width=600,
    height=800
)
fig.show()
```

```python
trend = df.groupby("age")["mental_health_risk_score"].mean().reset_index()

fig = px.line(
    trend,
    x="age",
    y="mental_health_risk_score",
    markers=True,
    title="Average Risk Score by Age"
)
fig.show()
```

```python
px.scatter(
	data_frame=df,
	x='depression_label',
	y='physical_activity',
	color='daily_social_media_hours',
	size='sleep_hours',
	hover_data=['sleep_hours']
)
```

```python
fig = px.box(
    df,
    x="gender",
    y="mental_health_risk_score",
    color="gender",
    title="Risk Score by Gender"
)
fig.show()
```

```python
fig = px.violin(
    df,
    x="depression_label",
    y="stress_level",
    color="depression_label",
    box=True
)
fig.show()
```

```python
trend = df.groupby("age")["daily_social_media_hours"].mean().reset_index()

fig = px.area(
    trend,
    x="age",
    y="daily_social_media_hours",
    title="Average Social Media Hours by Age"
)
fig.show()
```

```python
fig = px.sunburst(
    df,
    path=["gender","platform_usage","depression_label"],
    values="mental_health_risk_score"
)
fig.show()
```

```python
fig = px.density_heatmap(
    df,
    x="daily_social_media_hours",
    y="mental_health_risk_score"
)
fig.show()
```

```python
fig = px.scatter_matrix(
    df,
    dimensions=[
        "daily_social_media_hours",
    ],
    color="gender"
)
fig.show()
```
